```python
# Helper functions: Black-Scholes
def sim_jump_diffusion(X0, T, Ndt, mu, sigma, Nsims = 100, type = "BS", lam = 0, m = 0, delta = 0, p = 0.5, eta = [1, 1]):
    ## Basic parameters
    dt = T / Ndt # create time increment
    time = np.linspace(0, T, Ndt+1) # set the different times as a vector
    X = B = np.zeros((Ndt+1,Nsims)) # initialize matrix
    X[0,:] = X0 # initial value of the process

    ## Jump generation function
    drift_adj = 0
    def jump_distn(): return 0 # No jumps
    if type == "Merton":
        # Merton: Drift adjustment
        # drift_adj = lam * (np.exp(m + delta**2 / 2) - 1)

        # Generate jumps
        def jump_distn():
            # Number of jumps - Poisson process
            U = stats.uniform.rvs(size=Nsims)
            jump = U < (1 - np.exp(-lam * dt))  # Interarrival times exponentially dist
            # Jump size - Normal distribution
            jump_size = np.random.normal(loc=m, scale=delta, size=Nsims) # Jump size normally distributed
            return jump * jump_size

    elif type == "Kou":
        # Kou: Drift adjustment
        # drift_adj = lam * (np.exp(m + delta**2 / 2) - 1)

        def jump_distn():
            # Number of jumps - Poisson process
            U = stats.uniform.rvs(size=Nsims)
            jump = U < (1 - np.exp(-lam * dt))
            # Jump size - Double exponential
            B = np.random.binomial(n = 1, p = p, size = Nsims) # Bernoulli rv - whether jump is positive or negative
            jump_size = np.zeros(Nsims)
            for i in range(Nsims):
                jump_size[i] = (-1)**(B[i] + 1) * np.random.exponential(1/eta[B[i]]) # Jump size exponentially distributed
            return jump * jump_size


    ## Simulation
    for idx in range(Ndt):
        Z = np.random.randn(Nsims)
        # X[idx+1,:] = X[idx,:] + (r - sigma**2 / 2 - merton) * dt + sigma * np.sqrt(dt) * Z + jump_distn()
        X[idx+1,:] = X[idx,:] * np.exp((mu - sigma**2 / 2 - drift_adj) * dt + sigma * np.sqrt(dt) * Z + jump_distn())

    return time, X
```

```python
## Helper functions
# Black-Scholes
def call_BS(St, k, r, sigma, T, t = 0):
    dpos = (np.log(St/k) + (r + sigma**2/2)*(T - t))/(sigma * np.sqrt(T - t))
    dneg = (np.log(St/k) + (r - sigma**2/2)*(T - t))/(sigma * np.sqrt(T - t))
    return St * stats.norm.cdf(dpos) - k * np.exp(-r * (T - t)) * stats.norm.cdf(dneg)

# Merton
def call_Merton(lam, m, delta, St, k, r, sigma, T, t = 0):
    val = np.zeros(len(k))
    lam_bar = lam * np.exp(m + delta**2/2)
    delta_T = T - t
    # for i in range(100):
    for j in range(len(k)):
        i = 0
        diff = 1
        while diff > 0.00001:
            sigma_i = np.sqrt(sigma**2 + (i * delta**2)/delta_T)
            r_i = r - lam * (np.exp(m + delta**2/2) - 1) + i*(m + delta**2/2)/delta_T
            diff = (lam_bar * delta_T)**i/np.math.factorial(i) * call_BS(St, k[j], r_i, sigma_i, T, t)
            val[j] += diff
            i += 1
    return np.exp(-lam_bar * delta_T) * val


# Tests
print(call_BS(50, np.arange(5) + 1, 0.05, 0.2, 1))

print(call_Merton(1, 0.1, 0.1, 50, np.arange(5) + 1, 0.05, 0.2, 1))

round(3.33333, 2)
```

```python
## Helper functions
# Black-Scholes
def call_BS(St, k, r, sigma, T, t = 0):
    dpos = (np.log(St/k) + (r + sigma**2/2)*(T - t))/(sigma * np.sqrt(T - t))
    dneg = (np.log(St/k) + (r - sigma**2/2)*(T - t))/(sigma * np.sqrt(T - t))
    return St * stats.norm.cdf(dpos) - k * np.exp(-r * (T - t)) * stats.norm.cdf(dneg)

# Merton
def call_Merton(lam, m, delta, St, k, r, sigma, T, t = 0):
    val = np.zeros(len(St))
    lam_bar = lam * np.exp(m + delta**2/2)
    delta_T = T - t
    # for i in range(100):
    for j in range(len(St)):
        i = 0
        diff = 1
        while diff > 0.00001:
            sigma_i = np.sqrt(sigma**2 + (i * delta**2)/delta_T)
            r_i = r - lam * (np.exp(m + delta**2/2) - 1) + i*(m + delta**2/2)/delta_T
            diff = (lam_bar * delta_T)**i/np.math.factorial(i) * call_BS(St[j], k, r_i, sigma_i, T, t)
            val[j] += diff
            i += 1
    return np.exp(-lam_bar * delta_T) * val
```

```python
# Delta-Greek Approximation
def greek_delta(prices, end, start = 0):
    dt = (end - start)/(len(prices) - 1)
    dprices = np.zeros(len(prices))
    dprices[1:len(prices)] = prices[1:len(prices)] - prices[0:(len(prices) - 1)]
    dprices[0] = dprices[1]

    return dprices / dt

# Setup
fig, axes = plt.subplots(1, 2, figsize=(12,6), sharex=True)

# Constant diffusion
delta_greek1 = np.transpose(np.array([greek_delta(price1[:, 0], 100, 1), greek_delta(price1[:, 1], 100, 1)]))
# Plot the delta (greeks) approximation
axes[0].plot(St, delta_greek1, linewidth=lw, label = ["BS", "Merton"])
axes[0].vlines(x=k, ymin=0, ymax=1, colors = 'gray', linestyles = 'dotted')
axes[0].set_title("Constant Diffusion: " + r"sigma = " + str(sigma_BS))
axes[0].set_xlabel('Asset Price (St)')
axes[0].set_ylabel('Delta')
axes[0].legend()

# Adjusted diffusion
delta_greek2 = np.transpose(np.array([greek_delta(price2[:, 0], 100, 1), greek_delta(price2[:, 1], 100, 1)]))
# Plot the delta (greeks) approximation
axes[1].plot(St, delta_greek2, linewidth=lw, label = ["BS", "Merton"])
axes[1].vlines(x=k, ymin=0, ymax=1, colors = 'gray', linestyles = 'dotted')
axes[1].set_title("Adjusted Diffusion: " + r"sigmaMerton = " + str(round(sigma_Merton, 4)))
axes[1].set_xlabel('Asset Price (St)')
axes[1].set_ylabel('Delta')
axes[1].legend()

plt.show()
```
